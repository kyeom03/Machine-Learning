```python
import os
import random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras.layers import (
    Input, Conv1D, MaxPooling1D,
    GlobalAveragePooling1D,
    BatchNormalization, ReLU,
    Add, Dense, Dropout
)

from tensorflow.keras.models import Model
from tensorflow.keras.utils import to_categorical

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

from sklearn.utils.class_weight import compute_class_weight

from tensorflow.keras.layers import (
    Multiply,
    Concatenate,
    GlobalMaxPooling1D
)

from tensorflow.keras.layers import Reshape

# ============================================================
# 1. SEED
# ============================================================

SEED = 42

os.environ["PYTHONHASHSEED"] = str(SEED)

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


# ============================================================
# 2. CONFIGURATION
# ============================================================

DATA_PATH = "/content/drive/MyDrive/Colab Notebooks/WISDM_ar_v1.1_raw.txt"

MODEL_PATH = "/content/best_resnet_inception_cbam.keras"

WINDOW_SIZE = 128
STRIDE = 64

N_CHANNELS = 3
N_CLASSES = 6

TIME_GAP_THRESHOLD = 200_000_000

BATCH_SIZE = 64
EPOCHS = 50

LEARNING_RATE = 0.001

class_names = [
    "WALKING",
    "JOGGING",
    "UPSTAIRS",
    "DOWNSTAIRS",
    "SITTING",
    "STANDING"
]

activity_map = {
    "Walking": 0,
    "Jogging": 1,
    "Upstairs": 2,
    "Downstairs": 3,
    "Sitting": 4,
    "Standing": 5
}


# ============================================================
# 3. LOAD WISDM DATASET
# ============================================================

def load_wisdm(path):

    columns = [
        "user",
        "activity",
        "timestamp",
        "x",
        "y",
        "z"
    ]

    df = pd.read_csv(
        path,
        header=None,
        names=columns,
        sep=",",
        engine="python",
        on_bad_lines="skip",
        skipinitialspace=True
    )

    df["z"] = (
        df["z"]
        .astype(str)
        .str.replace(";", "", regex=False)
    )

    for col in ["user", "timestamp", "x", "y", "z"]:

        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        )

    df["activity"] = (
        df["activity"]
        .astype(str)
        .str.strip()
    )

    df = df.dropna(
        subset=columns
    ).copy()

    df["user"] = df["user"].astype(int)

    df["timestamp"] = df["timestamp"].astype(np.int64)

    df = df[
        df["activity"].isin(activity_map.keys())
    ].copy()

    df["label"] = (
        df["activity"]
        .map(activity_map)
        .astype(int)
    )

    df = df.sort_values(
        ["user", "timestamp"]
    ).reset_index(drop=True)

    return df


df = load_wisdm(DATA_PATH)

print("Dataset shape:", df.shape)

print("Number of users:", df["user"].nunique())

print("\nActivity distribution:")

print(df["activity"].value_counts())



# ============================================================
# 4. TIMESTAMP CHECK
# ============================================================

timestamp_diff = (
    df.groupby("user")["timestamp"].diff()
)

print("\nTimestamp statistics:")

print(timestamp_diff.describe())


# ============================================================
# 5. SUBJECT-WISE SPLIT
# ============================================================

users = np.sort(df["user"].unique())

rng = np.random.RandomState(SEED)

rng.shuffle(users)

n_users = len(users)

n_train = int(n_users * 0.70)

n_val = int(n_users * 0.15)

train_users = users[:n_train]

val_users = users[
    n_train:n_train + n_val
]

test_users = users[
    n_train + n_val:
]

print("\nTrain users:", train_users)

print("Validation users:", val_users)

print("Test users:", test_users)


assert len(set(train_users) & set(val_users)) == 0

assert len(set(train_users) & set(test_users)) == 0

assert len(set(val_users) & set(test_users)) == 0


# ============================================================
# 6. SPLIT DATASET
# ============================================================

train_df = df[
    df["user"].isin(train_users)
].copy()

val_df = df[
    df["user"].isin(val_users)
].copy()

test_df = df[
    df["user"].isin(test_users)
].copy()


for name, split_df in [
    ("TRAIN", train_df),
    ("VALIDATION", val_df),
    ("TEST", test_df)
]:

    print(f"\n{name} activity distribution:")

    print(split_df["activity"].value_counts())
```

```python
# ============================================================
# 7. SLIDING WINDOW GENERATION
# ============================================================

def create_windows(
    data,
    window_size=128,
    stride=64
):

    X = []
    y = []
    window_users = []

    for user, user_df in data.groupby("user"):

        user_df = user_df.sort_values(
            "timestamp"
        ).reset_index(drop=True)

        values = user_df[
            ["x", "y", "z"]
        ].to_numpy(dtype=np.float32)

        labels = user_df[
            "label"
        ].to_numpy(dtype=np.int32)

        timestamps = user_df[
            "timestamp"
        ].to_numpy(dtype=np.int64)

        # Activity boundary
        activity_change = (
            labels[1:] != labels[:-1]
        )

        # Timestamp boundary
        timestamp_diff = np.diff(timestamps)

        time_gap = (
            timestamp_diff > TIME_GAP_THRESHOLD
        )

        # Combine boundaries
        boundaries = np.where(
            activity_change | time_gap
        )[0] + 1

        starts = np.concatenate([
            [0],
            boundaries
        ])

        ends = np.concatenate([
            boundaries,
            [len(values)]
        ])

        # Generate windows
        for start, end in zip(starts, ends):

            segment_x = values[start:end]

            segment_y = labels[start:end]

            if len(segment_x) < window_size:
                continue

            for i in range(
                0,
                len(segment_x) - window_size + 1,
                stride
            ):

                window_x = segment_x[
                    i:i + window_size
                ]

                window_y = segment_y[
                    i:i + window_size
                ]

                label = window_y[0]

                X.append(window_x)

                y.append(label)

                window_users.append(user)

    return (
        np.asarray(X, dtype=np.float32),
        np.asarray(y, dtype=np.int32),
        np.asarray(window_users)
    )


# ============================================================
# 8. GENERATE WINDOWS
# ============================================================

X_train, y_train, train_window_users = create_windows(
    train_df,
    WINDOW_SIZE,
    STRIDE
)

X_val, y_val, val_window_users = create_windows(
    val_df,
    WINDOW_SIZE,
    STRIDE
)

X_test, y_test, test_window_users = create_windows(
    test_df,
    WINDOW_SIZE,
    STRIDE
)


print("\nWindow Shapes")

print("X_train:", X_train.shape)

print("X_val:", X_val.shape)

print("X_test:", X_test.shape)


if min(
    len(X_train),
    len(X_val),
    len(X_test)
) == 0:

    raise ValueError(
        "Empty window dataset. Check data and timestamp units."
    )


# ============================================================
# 9. CLASS DISTRIBUTION
# ============================================================

for name, labels in [
    ("TRAIN", y_train),
    ("VALIDATION", y_val),
    ("TEST", y_test)
]:

    print(f"\n{name} window distribution:")

    counts = np.bincount(
        labels,
        minlength=N_CLASSES
    )

    for i, activity in enumerate(class_names):

        print(
            f"{activity}: {counts[i]}"
        )


# Check all classes in training data

classes = np.unique(y_train)

if len(classes) != N_CLASSES:

    raise ValueError(
        "Some activity classes are missing from training data."
    )


# ============================================================
# 10. NORMALIZATION
# ============================================================

mean = X_train.mean(
    axis=(0, 1),
    keepdims=True
)

std = X_train.std(
    axis=(0, 1),
    keepdims=True
)

std = np.maximum(
    std,
    1e-8
)

X_train = (X_train - mean) / std

X_val = (X_val - mean) / std

X_test = (X_test - mean) / std


print("\nNormalization completed.")

print("Mean:", mean.reshape(-1))

print("Std:", std.reshape(-1))


# ============================================================
# 11. ONE-HOT ENCODING
# ============================================================

y_train_onehot = to_categorical(
    y_train,
    num_classes=N_CLASSES
)

y_val_onehot = to_categorical(
    y_val,
    num_classes=N_CLASSES
)

y_test_onehot = to_categorical(
    y_test,
    num_classes=N_CLASSES
)


# ============================================================
# 12. CLASS WEIGHTS
# ============================================================

weights = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(N_CLASSES),
    y=y_train
)

class_weights = dict(
    enumerate(weights)
)

print("\nClass weights:")

print(class_weights)
```

```python
# ============================================================
# 13. RESNET RESIDUAL BLOCK
# ============================================================

def residual_block(x, filters, kernel_size=3, dropout_rate=0.2):

    shortcut = x

    x = Conv1D(
        filters,
        kernel_size,
        padding="same",
        use_bias=False,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )(x)

    x = BatchNormalization()(x)
    x = ReLU()(x)

    x = Dropout(dropout_rate)(x)

    x = Conv1D(
        filters,
        kernel_size,
        padding="same",
        use_bias=False,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )(x)

    x = BatchNormalization()(x)

    if shortcut.shape[-1] != filters:

        shortcut = Conv1D(
            filters,
            1,
            padding="same",
            use_bias=False
        )(shortcut)

        shortcut = BatchNormalization()(shortcut)

    x = Add()([x, shortcut])

    x = ReLU()(x)

    return x


# ============================================================
# 14. INCEPTION RESIDUAL BLOCK
# ============================================================

def inception_residual_block(x, filters, dropout_rate=0.2):

    shortcut = x

    branch1 = Conv1D(
        filters // 3,
        3,
        padding="same",
        use_bias=False,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )(x)

    branch1 = BatchNormalization()(branch1)
    branch1 = ReLU()(branch1)

    branch2 = Conv1D(
        filters // 3,
        5,
        padding="same",
        use_bias=False,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )(x)

    branch2 = BatchNormalization()(branch2)
    branch2 = ReLU()(branch2)

    branch3 = Conv1D(
        filters // 3,
        7,
        padding="same",
        use_bias=False,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )(x)

    branch3 = BatchNormalization()(branch3)
    branch3 = ReLU()(branch3)

    x = Concatenate()([
        branch1,
        branch2,
        branch3
    ])

    x = Conv1D(
        filters,
        1,
        padding="same",
        use_bias=False,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )(x)

    x = BatchNormalization()(x)

    if shortcut.shape[-1] != filters:

        shortcut = Conv1D(
            filters,
            1,
            padding="same",
            use_bias=False,
            kernel_regularizer=tf.keras.regularizers.l2(1e-4)
        )(shortcut)

        shortcut = BatchNormalization()(shortcut)

    x = Add()([x, shortcut])

    x = ReLU()(x)

    x = Dropout(dropout_rate)(x)

    return x

# ============================================================
# 15. CUSTOM TEMPORAL POOLING LAYER
# ============================================================

@tf.keras.utils.register_keras_serializable(
    package="HAR"
)
class TemporalAveragePooling(tf.keras.layers.Layer):

    def call(self, inputs):

        return tf.reduce_mean(
            inputs,
            axis=-1,
            keepdims=True
        )


@tf.keras.utils.register_keras_serializable(
    package="HAR"
)
class TemporalMaxPooling(tf.keras.layers.Layer):

    def call(self, inputs):

        return tf.reduce_max(
            inputs,
            axis=-1,
            keepdims=True
        )


# ============================================================
# 16. CHANNEL + TEMPORAL ATTENTION
# ============================================================

def attention_block_1d(x, reduction=8):

    channels = int(x.shape[-1])

    reduced = max(channels // reduction, 1)

    # --------------------------------
    # CHANNEL ATTENTION
    # --------------------------------

    avg_pool = GlobalAveragePooling1D()(x)

    max_pool = GlobalMaxPooling1D()(x)

    shared_dense1 = Dense(
        reduced,
        activation="relu"
    )

    shared_dense2 = Dense(
        channels
    )

    avg_out = shared_dense2(
        shared_dense1(avg_pool)
    )

    max_out = shared_dense2(
        shared_dense1(max_pool)
    )

    channel_attention = Add()([
        avg_out,
        max_out
    ])

    channel_attention = tf.keras.layers.Activation(
        "sigmoid"
    )(channel_attention)

    channel_attention = Reshape(
        (1, channels)
    )(channel_attention)

    x = Multiply()([
        x,
        channel_attention
    ])

    # --------------------------------
    # TEMPORAL ATTENTION
    # --------------------------------

    avg_temporal = TemporalAveragePooling()(x)

    max_temporal = TemporalMaxPooling()(x)

    temporal_features = Concatenate(
        axis=-1
    )([
        avg_temporal,
        max_temporal
    ])

    temporal_attention = Conv1D(
        1,
        7,
        padding="same",
        activation="sigmoid",
        use_bias=False,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )(temporal_features)

    x = Multiply()([
        x,
        temporal_attention
    ])

    return x


# ============================================================
# 17. BUILD RESNET + INCEPTION + ATTENTION
# ============================================================

def build_hybrid_model():

    inputs = Input(
        shape=(WINDOW_SIZE, N_CHANNELS)
    )

    # --------------------------------
    # INITIAL CONV
    # --------------------------------

    x = Conv1D(
        32,
        5,
        padding="same",
        use_bias=False,
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )(inputs)

    x = BatchNormalization()(x)
    x = ReLU()(x)

    # --------------------------------
    # RESNET BLOCK 1
    # --------------------------------

    x = residual_block(
        x,
        32,
        dropout_rate=0.2
    )

    x = MaxPooling1D(2)(x)

    # --------------------------------
    # INCEPTION RESIDUAL BLOCK
    # --------------------------------

    x = inception_residual_block(
        x,
        64,
        dropout_rate=0.2
    )

    x = MaxPooling1D(2)(x)

    # --------------------------------
    # CHANNEL + TEMPORAL ATTENTION
    # --------------------------------

    x = attention_block_1d(
        x,
        reduction=8
    )

    # --------------------------------
    # GLOBAL AVERAGE POOLING
    # --------------------------------

    x = GlobalAveragePooling1D()(x)

    # --------------------------------
    # DENSE
    # --------------------------------

    x = Dense(
        64,
        activation="relu",
        kernel_regularizer=tf.keras.regularizers.l2(1e-4)
    )(x)

    x = Dropout(0.5)(x)

    # --------------------------------
    # OUTPUT
    # --------------------------------

    outputs = Dense(
        N_CLASSES,
        activation="softmax"
    )(x)

    model = Model(
        inputs,
        outputs,
        name="WISDM_Improved_ResNet_Inception_Attention"
    )

    return model


# ============================================================
# 18. COMPILE MODEL
# ============================================================

model = build_hybrid_model()

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0003
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)

model.summary()


# ============================================================
# 19. CALLBACKS
# ============================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True,
    verbose=1
)

checkpoint = ModelCheckpoint(
    MODEL_PATH,
    monitor="val_loss",
    save_best_only=True,
    mode="min",
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=4,
    min_lr=1e-6,
    verbose=1
)


# ============================================================
# 20. MODEL TRAINING
# ============================================================

history = model.fit(

    X_train,

    y_train_onehot,

    validation_data=(
        X_val,
        y_val_onehot
    ),

    epochs=EPOCHS,

    batch_size=BATCH_SIZE,

    shuffle=True,

    class_weight=class_weights,

    callbacks=[
        early_stopping,
        reduce_lr,
        checkpoint
    ],

    verbose=1
)
```

```python
# ============================================================
# 21. TRAINING HISTORY
# ============================================================

plt.figure(figsize=(14, 5))

# Accuracy
plt.subplot(1, 2, 1)

plt.plot(
    history.history["accuracy"],
    label="Train Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("RESNET + INCEPTION + CBAM Accuracy")

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.legend()

plt.grid(True)


# Loss
plt.subplot(1, 2, 2)

plt.plot(
    history.history["loss"],
    label="Train Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("RESNET + INCEPTION + CBAM Loss")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()

# ============================================================
# 22. LOAD BEST MODEL
# ============================================================

model = tf.keras.models.load_model(
    MODEL_PATH
)


# ============================================================
# 23. TEST PREDICTION
# ============================================================

y_prob = model.predict(
    X_test,
    batch_size=BATCH_SIZE,
    verbose=0
)

y_pred = np.argmax(
    y_prob,
    axis=1
)


# ============================================================
# 24. EVALUATION METRICS
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)


print("\n================================")

print("RESNET + INCEPTION + CBAM TEST RESULTS")

print("================================")

print(f"Accuracy  : {accuracy:.4f}")

print(f"Precision : {precision:.4f}")

print(f"Recall    : {recall:.4f}")

print(f"Macro F1  : {f1:.4f}")


# ============================================================
# 25. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report")

print(
    classification_report(
        y_test,
        y_pred,
        labels=np.arange(N_CLASSES),
        target_names=class_names,
        digits=4,
        zero_division=0
    )
)


# ============================================================
# 26. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=np.arange(N_CLASSES)
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title("RESNET + INCEPTION + CBAM Confusion Matrix")

plt.xlabel("Predicted Label")

plt.ylabel("True Label")

plt.tight_layout()

plt.show()


# ============================================================
# 27. SAVE RESULTS
# ============================================================

results = {
    "Model": "RESNET + INCEPTION + CBAM",
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "Macro F1": f1
}

results_df = pd.DataFrame([results])

print("\nFinal Results")

print(results_df.to_string(index=False))
```
