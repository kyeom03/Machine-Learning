# ***1주차: 데이터처리 개요***
## ****데이터처리란****
- 원시데이터를 가공하여 사용 목적에 맞게 적재하는 일련의 과정

## ***데이터처리를 해야 하는 이유***
- 세상 모든 데이터가 우리의 최종 목적에 맞게 제공되지는 않습니다
- 따라서 그 데이터를 가공해야 합니다
  - 스프레드시트, 컴퓨터 프로그래밍, SQL 등등
- 가공할 데이터는 어디선가 가져와야 합니다
  - 다양한 출처로부터 수집해야 합니다
- 가공된 데이터는 어디엔가 저장해야 합니다
  - 다시 활용하기에 편리한 형태로 저장해야 합니다

## ***데이터를 제공하는 곳***
- 데이터 제공 플랫폼으로부터 제공받을 수 있습니다
  - https://www.data.go.kr 에 접속하여 어떤 데이터가 있는지 확인해봅시다
  - 이외에도 여러 데이터 제공 플랫폼이 있습니다

- 조직 내에서 데이터를 제공받을 수도 있습니다
- 업무적 협력관계에 있는 조직이나 기관으로부터 데이터를 제공받을 수도 있습니다
- 단순히 지인을 통해 데이터를 제공받을 수도 있습니다
- 이것이 의미하는 것은 데이터의 제공 형태는 다양할 수 있다는 것입니다
- 또한 우리가 원하는 형태로 최종 제공되지 않는 경우가 많다는 것입니다

## ***데이터를 가공하는 방법***
### ***스프레드시트***
- 소용량 데이터에서는 아주 효과적인 방법입니다
- 그러나 용량의 한계가 명확합니다

### ***SQL***
- 원시데이터가 DBMS의 한 테이블 형태로 제공된다면 유용한 방법입니다
- 만약 그렇지않다면 데이터 가공을 SQL로 시작하기는 어렵습니다

### ***컴퓨터 프로그래밍***
- 다양한 라이브러리를 이용해 유연하게 대처할 수 있습니다
- 라이브러리 사용방법을 익혀야 하는 번거로움은 있습니다
  - 어짜피 스프레드시트도 쓰려면 함수 배워야 합니다
  - 어짜피 SQL도 쓰려면 문법 배워야 합니다
  - 그래서 우리는 몇가지 라이브러리를 지정하고 이를 이용해 데이터처리프로그래밍을 익히도록 합니다

## ***ETL***
- Extract: 수집
- Transform: 변환
- Load: 적재

### ***ETL의 순서***
- 반드시 E -> T -> L로만 가야만 할까요?
- E -> L -> T -> L도 가능합니다
  - 원시데이터의 보존
  - 데이터 변환의 비가역성
  - 카메라에서 raw파일을 저장하는 이유
- 다시 말하면 굳이 ETL 순서에 맞게 배우는 것을 고집할 필요는 없어보입니다

### ***우리가 ETL을 배우는 순서***
- T부터 갑니다
- T를 먼저 배우고 앞뒤로 E와 L을 연계하는 형태로 진행하겠습니다

## ***데이터 활용 예시***
- opensky-network.org 의 데이터를 활용한 예시입니다
- 데이터를 1회성으로 수집하여 해당 시점의 항공기 추적 현황을 지도위에 시각화 합니다
- 편명에 대한 상세 운항 현황정보를 추가하고 이를 누적한 데이터를 구축하고자 합니다
- 이를 위해 필요한 것을 생각해봅니다
  - 1회성으로 수집한 데이터는 python 객체로 들어옵니다
  - 이를 어떻게 누적하여 저장할까요?
  - 항공편의 상세 운항 현황을 제공하는 곳이 있을까요?
    - https://www.data.go.kr/data/15095074/openapi.do 를 확인해봅니다
  - 이 데이터에는 항공기 편명이 ICAO 규격으로 제공되고 IATA 규격으로는 제공되지 않습니다
    - ICAO코드에 대응되는 IATA코드는 어떻게 알 수 있을까요?
        - https://openflights.org/data.php 를 확인해봅니다

````python
import requests

url = "https://opensky-network.org/api/states/all"

params = {
    "lamin": 37.0,
    "lomin": 125.5,
    "lamax": 37.9,
    "lomax": 127.5
}

response = requests.get(url, params=params)
data = response.json()

print("기준 시각:", data["time"])
print("항공기 수:", len(data["states"]))
````

````python
import pandas as pd

# https://opensky-network.org/api/states/all 에서 제공되는 데이터 컬럼 정보 입니다
columns = [
    "icao24",
    "callsign",
    "origin_country",
    "time_position",
    "last_contact",
    "longitude",
    "latitude",
    "baro_altitude",
    "on_ground",
    "velocity",
    "true_track",
    "vertical_rate",
    "sensors",
    "geo_altitude",
    "squawk",
    "spi",
    "category"
]

df = pd.DataFrame(data["states"], columns=columns)

df.head()
````

````python
import folium
import math

m = folium.Map(
    location=[37.5, 126.9],
    zoom_start=8
)

for _, row in df.dropna(
    subset=["latitude", "longitude", "true_track", "velocity"]
).iterrows():

    lat = row["latitude"]
    lon = row["longitude"]
    track = row["true_track"]

    # 화살표 길이를 적당히 표현하기 위해 조정하는 식입니다
    length = 0.05 * (row["velocity"] / 300)

    # 북쪽을 기준으로 0°, track값은 시계방향으로 증가합니다
    # 이러한 정보는 어떻게 얻는 것일까요? https://openskynetwork.github.io/opensky-api/rest.html 에서 확인합니다
    # 지도상의 위도/경도 방향으로 변환
    angle = math.radians(track)

    end_lat = lat + length * math.cos(angle)
    end_lon = lon + length * math.sin(angle)

    # 항공기 위치
    folium.CircleMarker(
        location=[lat, lon],
        radius=5,
        popup=(
            f"Callsign: {row['callsign']}<br>"
            f"Altitude: {row['baro_altitude']} m<br>"
            f"Velocity: {row['velocity']:.1f} m/s<br>"
            f"Track: {track:.1f}°<br>"
            f"Vertical rate: {row['vertical_rate']} m/s"
        ),
        fill=True
    ).add_to(m)

    # 진행 방향
    folium.PolyLine(
        locations=[
            [lat, lon],
            [end_lat, end_lon]
        ],
        weight=3
    ).add_to(m)

m
````
